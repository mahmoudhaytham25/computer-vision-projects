```python
def get_data_paths(path):
  paths= []
  labels = os.listdir(path)
  for label in labels:
    label_path= os.path.join(path, label)
    for img in os.listdir(label_path):
      img_path= os.path.join(label_path, img)
      paths.append((img_path, label))
  return paths
```

```python
train_paths= get_data_paths(train_path)
val_paths= get_data_paths(val_path)
test_paths= get_data_paths(test_path)
```

```python
train_paths= np.array(train_paths)
val_paths= np.array(val_paths)
test_paths= np.array(test_paths)
```

```python
def map_label(label):
  if label == 'happy':
    return 0
  elif label == 'Sad':
    return 1
  elif label == 'Angry':
    return 2
  else:
    return 3
```

```python
def demap_label(label):
    if label == 0:
        return 'happy'
    elif label == 1:
        return 'Sad'
    elif label == 2:
        return 'Angry'
    else:
        return 'Other'
```

```python
class ImgDataset(Dataset):
    def __init__(self, paths, transform= None):
      self.paths= paths[:, 0]
      self.labels= paths[:, 1]
      self.transform= transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        img_path = self.paths[idx]
        label = self.labels[idx]
        image = Image.open(img_path).convert('RGB')
        if self.transform:
            image = self.transform(image)
        return image, map_label(label)
```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),       # Resize the image to 224x224
    transforms.ToTensor(),              # Convert PIL Image to PyTorch tensor
    #transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]) # Normalize
])
```

```python
train_dataset= ImgDataset(train_paths, transform= transform)
val_dataset= ImgDataset(val_paths, transform= transform)
test_dataset= ImgDataset(test_paths, transform= transform)
```

```python
batch_size= 16
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
# Function to display a batch of images and labels
def show_images(images, labels, num_images=16):
    """
    Displays images in a grid with their labels.
    Args:
    - images: Batch of images (Tensor).
    - labels: Corresponding labels (Tensor or list).
    - num_images: Number of images to display.
    """
    images = images[:num_images]  # Select the number of images to display
    labels = labels[:num_images]  # Select corresponding labels

    # Determine grid size
    grid_size = int(num_images ** 0.5)
    fig, axs = plt.subplots(grid_size, grid_size, figsize=(8, 8))

    for i, ax in enumerate(axs.flatten()):
        if i < len(images):
            # Convert tensor to a numpy array for visualization
            image = images[i].permute(1, 2, 0).cpu().numpy()  # CHW -> HWC
            ax.imshow(image)
            ax.set_title(f"Label: {demap_label(labels[i])}")
            ax.axis('off')
        else:
            ax.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
def augment_imgs(imgs, labels):
    augmented_imgs = []
    augmented_labels = []

    for img, label in zip(imgs, labels):
        # Original image
        augmented_imgs.append(img)
        augmented_labels.append(label)

        # Horizontal flip
        img_hflip = TF.hflip(img)
        augmented_imgs.append(img_hflip)
        augmented_labels.append(label)

        # Vertical flip
        img_vflip = TF.vflip(img)
        augmented_imgs.append(img_vflip)
        augmented_labels.append(label)

        # Combined flip (horizontal + vertical)
        img_hvflip = TF.hflip(img_vflip)
        augmented_imgs.append(img_hvflip)
        augmented_labels.append(label)

    # Convert lists to tensors
    augmented_imgs = torch.stack(augmented_imgs)
    augmented_labels = torch.tensor(augmented_labels)

    return augmented_imgs, augmented_labels
```

# Model

```python
class DenseNet121(nn.Module):
    def __init__(self, num_classes=4):
        super(DenseNet121, self).__init__()

        # Initial convolution
        self.initial = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=7, stride=2, padding=3, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1),
        )

        # Dense and transition blocks
        self.dense1 = self._make_dense_block(64, 32, 6)
        self.trans1 = self._transition_layer(64 + 6 * 32, 128)

        self.dense2 = self._make_dense_block(128, 32, 12)
        self.trans2 = self._transition_layer(128 + 12 * 32, 256)

        self.dense3 = self._make_dense_block(256, 32, 24)
        self.trans3 = self._transition_layer(256 + 24 * 32, 512)

        self.dense4 = self._make_dense_block(512, 32, 16)

        # Classification layer
        self.classifier = nn.Sequential(
            nn.Dropout(),
            nn.Linear(512 + 16 * 32, num_classes),
        )

        # Initialize weights
        self._initialize_weights()

    def _make_dense_block(self, in_channels, growth_rate, num_layers):
        layers = nn.ModuleList()
        for i in range(num_layers):
            layers.append(self._dense_layer(in_channels + i * growth_rate, growth_rate))
        return layers

    def _dense_layer(self, in_channels, growth_rate):
        return nn.Sequential(
            nn.BatchNorm2d(in_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(in_channels, growth_rate, kernel_size=3, padding=1, bias=False),
            nn.Dropout(),
        )

    def _transition_layer(self, in_channels, out_channels):
        return nn.Sequential(
            nn.BatchNorm2d(in_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(in_channels, out_channels, kernel_size=1, bias=False),
            nn.Dropout(),
            nn.AvgPool2d(kernel_size=2, stride=2),
        )

    def forward_dense_block(self, x, layers):
        outputs = [x]  # Start with the initial input feature map
        for layer in layers:
            new_feature = layer(torch.cat(outputs, dim=1))  # Concatenate along the channel dimension
            outputs.append(new_feature)
        return torch.cat(outputs, dim=1)  # Concatenate all feature maps at the end

    def forward(self, x):
        # Initial convolution and pooling
        x = self.initial(x)

        # Dense and transition blocks
        x = self.forward_dense_block(x, self.dense1)
        x = self.trans1(x)

        x = self.forward_dense_block(x, self.dense2)
        x = self.trans2(x)

        x = self.forward_dense_block(x, self.dense3)
        x = self.trans3(x)

        x = self.forward_dense_block(x, self.dense4)

        # Global average pooling and classification
        x = nn.AdaptiveAvgPool2d((1, 1))(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)

        return x

    def _initialize_weights(self):
        # Apply Xavier initialization to convolutional layers and fully connected layers
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    init.zeros_(m.bias)
            elif isinstance(m, nn.Linear):
                init.xavier_normal_(m.weight)
                if m.bias is not None:
                    init.zeros_(m.bias)
            elif isinstance(m, nn.BatchNorm2d):
                init.ones_(m.weight)
                init.zeros_(m.bias)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = DenseNet121(num_classes=4).to(device)
```

```python
print(model)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.0001)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=3, threshold= 1e-3)
```

```python
num_epochs= 30
train_losses=[]
train_accuracies=[]
val_losses=[]
val_accuracies=[]
# Training loop
for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0
    correct_train = 0
    total_train = 0

    # Training step
    for images, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}"):
        images, labels= augment_imgs(images, labels)
        images, labels = images.to(device), labels.to(device)  # Move data to device

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()


        train_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        correct_train += (predicted == labels).sum().item()
        total_train += labels.size(0)
    
    train_accuracy = 100 * correct_train / total_train
    train_losses.append(train_loss/len(train_loader))
    train_accuracies.append(train_accuracy) 
    print(f"Training Loss: {train_loss/len(train_loader):.4f}, Training Accuracy: {train_accuracy:.2f}%")

    # Validation step
    model.eval()
    val_loss = 0.0
    correct_val = 0
    total_val = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)

            # Forward pass
            outputs = model(images)
            loss = criterion(outputs, labels)

            # Track loss and accuracy
            val_loss += loss.item()
            _, predicted = torch.max(outputs, 1)
            correct_val += (predicted == labels).sum().item()
            total_val += labels.size(0)

    scheduler.step(val_loss/len(val_loader))
    val_accuracy = 100 * correct_val / total_val
    val_losses.append(val_loss/len(val_loader))
    val_accuracies.append(val_accuracy) 
    print(f"Validation Loss: {val_loss/len(val_loader):.4f}, Validation Accuracy: {val_accuracy:.2f}%")
    current_lr = optimizer.param_groups[0]['lr']
    print(f"Current Learning Rate: {current_lr}")
    print()
    torch.save(model.state_dict(), f"DenseNet121_epoch{epoch+1}_trained.pth")
```

```python
print(train_losses)
print(train_accuracies)
print(val_losses)
print(val_accuracies)
```
